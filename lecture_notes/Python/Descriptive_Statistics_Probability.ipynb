## 📋 Our Sample Data

```python
marks = [55, 70, 85, 60, 90, 75, 80, 65, 70, 95, 50, 72, 68, 88, 76]
marks = [6,7,8,4,3,7,6,1,7,2,4,7,6,2]
print("Marks:", marks)
print("Total students:", len(marks))
```

# 📍 PART 2: Measures of Central Tendency

These tell us the **center** of the data.

| Measure | What it means |
|---|---|
| **Mean** | Average of all values |
| **Median** | Middle value when sorted |
| **Mode** | Most repeated value |

## 1️⃣ Mean (Average)
**Formula:** Mean = Sum of all values ÷ Count

```python
mean = np.mean(marks)
print(f"Mean = {mean}")
```

Manual calculation of mean:

```python
manual_mean = sum(marks) / len(marks)
print(f"Manual Mean = {manual_mean}")
```

## 2️⃣ Median (Middle Value)
Sort the data → pick the middle number.

```python
median = np.median(marks)
print(f"Sorted: {sorted(marks)}")
print(f"Median = {median}")
```

## 3️⃣ Mode (Most Frequent Value)

```python
mode = stats.mode(marks)
print(f"Mode = {mode}")
```

## All Three Together

```python
print(f"Mean   = {np.mean(marks):.2f}")
print(f"Median = {np.median(marks)}")
print(f"Mode   = {stats.mode(marks)}")
```

## 🤔 When to Use Which?

| Use | When |
|---|---|
| **Mean** | No extreme values (outliers) |
| **Median** | Data has outliers (e.g. salary) |
| **Mode** | Categorical data (e.g. favorite color) |

## Effect of an Outlier on Mean

```python
salaries = [30000, 32000, 31000, 29000, 33000, 500000]  # 500000 is outlier

print(f"Mean   = {np.mean(salaries):,.0f}")
print(f"Median = {np.median(salaries):,.0f}")
print("Median is better here! ✅")
```

# 📏 PART 3: Measures of Dispersion

These tell us how **spread out** the data is.

| Measure | What it means |
|---|---|
| **Range** | Max - Min |
| **Variance** | Average of squared differences from mean |
| **Std Dev** | Square root of variance |
| **CV** | Std Dev ÷ Mean × 100% |

## 4️⃣ Range

```python
data_range = max(marks) - min(marks)
print(f"Max = {max(marks)}, Min = {min(marks)}")
print(f"Range = {data_range}")
```

## 5️⃣ Variance
Measures how far each value is from the mean.

- **Population Variance** → `ddof=0` (whole data)
- **Sample Variance** → `ddof=1` (part of data)

```python
pop_var = np.var(marks, ddof=0)
sam_var = np.var(marks, ddof=1)

print(f"Population Variance = {pop_var:.2f}")
print(f"Sample Variance     = {sam_var:.2f}")
```

## 6️⃣ Standard Deviation
Same unit as data. More useful than variance.

```python
std = np.std(marks, ddof=1)
print(f"Standard Deviation = {std:.2f}")
print(f"Most marks fall between {np.mean(marks)-std:.1f} and {np.mean(marks)+std:.1f}")
```

## 7️⃣ Coefficient of Variation (CV)
Compares spread between different datasets. Lower = more consistent.

```python
cv = (np.std(marks, ddof=1) / np.mean(marks)) * 100
print(f"CV = {cv:.2f}%")
```

## 8️⃣ Skewness
Tells us if data is **tilted** left or right.

| Value | Shape |
|---|---|
| `= 0` | Perfectly symmetric |
| `> 0` | Right skewed (tail on right) |
| `< 0` | Left skewed (tail on left) |

```python
from scipy import stats as sp

skewness = sp.skew(marks)
print(f"Skewness = {skewness:.4f}")
```

## 9️⃣ Kurtosis
Measures **sharpness** of the peak.

| Value | Meaning |
|---|---|
| `= 0` | Normal (Mesokurtic) |
| `> 0` | Sharp peak (Leptokurtic) |
| `< 0` | Flat peak (Platykurtic) |

```python
kurtosis = sp.kurtosis(marks)
print(f"Kurtosis = {kurtosis:.4f}")
```

# 📌 PART 4: Measures of Position

These tell us **where** a value stands in the data.

| Measure | What it does |
|---|---|
| **Z-Score** | How many std devs from mean |
| **Percentile** | What % of data falls below |
| **Quartile** | Splits data into 4 equal parts |

## 🔟 Z-Score
**Formula:** Z = (value - mean) ÷ std_dev

- Z = 0 → value is at the mean
- Z = +2 → 2 std devs above mean
- Z = -1 → 1 std dev below mean

```python
mean = np.mean(marks)
std  = np.std(marks, ddof=1)

z_scores = [(x - mean) / std for x in marks]

for m, z in zip(marks, z_scores):
    print(f"Mark={m}  Z-Score={z:.2f}")
```

Find Z-score for a single value:

```python
score = 90
z = (score - mean) / std
print(f"Z-score of {score} = {z:.2f}")
```

## 1️⃣1️⃣ Percentile
90th percentile = 90% of students scored **below** this value.

```python
p25 = np.percentile(marks, 25)
p50 = np.percentile(marks, 50)
p75 = np.percentile(marks, 75)
p90 = np.percentile(marks, 90)

print(f"25th Percentile = {p25}")
print(f"50th Percentile = {p50}  (same as median)")
print(f"75th Percentile = {p75}")
print(f"90th Percentile = {p90}")
```

## 1️⃣2️⃣ Quartiles
Splits data into **4 equal parts**.

| Quartile | Percentile |
|---|---|
| Q1 | 25th |
| Q2 | 50th (Median) |
| Q3 | 75th |

```python
Q1 = np.percentile(marks, 25)
Q2 = np.percentile(marks, 50)
Q3 = np.percentile(marks, 75)
IQR = Q3 - Q1

print(f"Q1  = {Q1}")
print(f"Q2  = {Q2}")
print(f"Q3  = {Q3}")
print(f"IQR = {IQR}  (Interquartile Range)")
```

## 1️⃣3️⃣ Detect Outliers using IQR
Values outside `Q1 - 1.5×IQR` or `Q3 + 1.5×IQR` are **outliers**.

```python
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = [x for x in marks if x < lower or x > upper]

print(f"Lower fence = {lower}")
print(f"Upper fence = {upper}")
print(f"Outliers    = {outliers}")
```

# 🔗 PART 5: Correlation

Correlation measures the **relationship** between two variables.

| Value | Meaning |
|---|---|
| `+1` | Perfect positive |
| `-1` | Perfect negative |
| `0`  | No relationship |

## 1️⃣4️⃣ Correlation Coefficient

```python
study_hours = [2, 3, 5, 1, 6, 4, 7, 3, 5, 8, 1, 4, 3, 6, 5]
test_scores = [55, 70, 85, 60, 90, 75, 80, 65, 70, 95, 50, 72, 68, 88, 76]

corr = np.corrcoef(study_hours, test_scores)[0, 1]
print(f"Correlation = {corr:.4f}")
```

Plot the relationship:

```python
plt.scatter(study_hours, test_scores, color='teal')
plt.xlabel("Study Hours")
plt.ylabel("Test Scores")
plt.title("Study Hours vs Test Scores")
plt.show()
```

# 🎲 PART 7: Probability

Probability = how likely an event will happen.

**Formula:**
$$P(A) = \frac{\text{Favorable outcomes}}{\text{Total outcomes}}$$

- Always between **0** (impossible) and **1** (certain)

## 1️⃣9️⃣ Basic Probability Example
Toss a coin → P(Head) = ?

```python
p_head = 1 / 2
print(f"P(Head) = {p_head}")

p_six = 1 / 6
print(f"P(Rolling a 6) = {p_six:.4f}")
```

---
## 🔵 Types of Probability

| Type | Description | Example |
|---|---|---|
| **Joint** | P(A and B) | P(Rain AND Cold) |
| **Marginal** | P(A) alone | P(Rain) |
| **Conditional** | P(A given B) | P(Rain \| Cloudy) |

## 2️⃣0️⃣ Joint Probability — P(A and B)
Both events happen together. For independent events: P(A∩B) = P(A) × P(B)

```python
p_head  = 0.5
p_six   = 1/6

p_both = p_head * p_six
print(f"P(Head AND Six) = {p_both:.4f}")
```

## 2️⃣1️⃣ Marginal Probability
Probability of one event regardless of others.

```python
# 100 students: 60 like cricket, 40 like football
total = 100
cricket = 60
football = 40

p_cricket  = cricket / total
p_football = football / total

print(f"P(Cricket)  = {p_cricket}")
print(f"P(Football) = {p_football}")
```

## 2️⃣2️⃣ Conditional Probability — P(A | B)
Probability of A **given** B has happened.

**Formula:** P(A|B) = P(A ∩ B) ÷ P(B)

```python
# P(pass given studied)
# Out of 100 students:
# 60 studied, 50 studied and passed

p_studied_and_passed = 50 / 100
p_studied = 60 / 100

p_pass_given_studied = p_studied_and_passed / p_studied
print(f"P(Pass | Studied) = {p_pass_given_studied:.2f}")
```

## 2️⃣3️⃣ Bayes' Theorem
Update our belief when we get new information.

**Formula:**
$$P(A|B) = \frac{P(B|A) \times P(A)}{P(B)}$$

```python
# P(Disease | Positive Test)

p_disease  = 0.01   # 1% have disease
p_pos_given_disease   = 0.95  # test correct when sick
p_pos_given_no_disease = 0.05  # false alarm rate

p_no_disease = 1 - p_disease
p_positive = (p_pos_given_disease * p_disease) + (p_pos_given_no_disease * p_no_disease)

p_disease_given_pos = (p_pos_given_disease * p_disease) / p_positive

print(f"P(Disease | Positive Test) = {p_disease_given_pos:.4f}")
print(f"That's only {p_disease_given_pos*100:.2f}% — surprising! 🤔")
```

# 📉 PART 8: Probability Distributions

A distribution shows all possible values and how likely each one is.

| Type | Examples |
|---|---|
| **Discrete** | Binomial, Poisson |
| **Continuous** | Normal, Uniform, Exponential |

## 2️⃣4️⃣ Uniform Distribution
Every value has **equal probability** (like rolling a die).

```python
data = np.random.uniform(0, 10, 1000)

plt.hist(data, bins=20, color='coral', edgecolor='black')
plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## 2️⃣5️⃣ Normal Distribution (Bell Curve) 🔔
Most common in real life. Symmetric around the mean.

- Mean, Median, Mode are **all equal**
- 68% data within ±1 std dev
- 95% data within ±2 std dev
- 99.7% data within ±3 std dev

```python
data = np.random.normal(loc=70, scale=10, size=1000)

plt.hist(data, bins=30, color='steelblue', edgecolor='black')
plt.title("Normal Distribution (Mean=70, Std=10)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## 2️⃣6️⃣ The 68–95–99.7 Rule

```python
data = np.random.normal(70, 10, 10000)
mean = np.mean(data)
std  = np.std(data)

within_1 = np.sum((data > mean-std) & (data < mean+std)) / len(data) * 100
within_2 = np.sum((data > mean-2*std) & (data < mean+2*std)) / len(data) * 100
within_3 = np.sum((data > mean-3*std) & (data < mean+3*std)) / len(data) * 100

print(f"Within ±1 std: {within_1:.1f}%  (theory: 68%)")
print(f"Within ±2 std: {within_2:.1f}%  (theory: 95%)")
print(f"Within ±3 std: {within_3:.1f}%  (theory: 99.7%)")
```

## 2️⃣7️⃣ Binomial Distribution (Discrete)
Used when there are only **two outcomes**: success or failure.

Example: Toss a coin 10 times. How many heads?

```python
from scipy.stats import binom

n = 10    # trials
p = 0.5   # probability of head

x = np.arange(0, 11)
y = binom.pmf(x, n, p)

plt.bar(x, y, color='mediumseagreen')
plt.title("Binomial Distribution (n=10, p=0.5)")
plt.xlabel("Number of Heads")
plt.ylabel("Probability")
plt.show()
```

## 2️⃣8️⃣ Poisson Distribution (Discrete)
Counts **events in a fixed time**. Example: calls per hour at a helpline.

```python
from scipy.stats import poisson

mu = 5   # average 5 calls per hour
x  = np.arange(0, 15)
y  = poisson.pmf(x, mu)

plt.bar(x, y, color='mediumpurple')
plt.title("Poisson Distribution (λ=5)")
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.show()
```
