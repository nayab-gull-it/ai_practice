```python
def twosum(nums):
    x = len(nums)
    for i in range(x):
        for j in range(i + 1, x):
            if nums[i] + nums[j] == 9:
                print("index are : ", i, j)
                
```

```python
twosum([8, 2, 4, 7, 5])
```

```python
def two_sums(nums , target):
    seen ={}
    for i in range(len(nums)):
        complement = target - nums[i]

        if complement in seen:
            return[seen[complement], i]
      

    seen[nums[i]] = i

    return []
```

```python
%%time 
result = two_sums([8, 2, 4, 7, 5], 9)
print("Indices are:", result)
```
