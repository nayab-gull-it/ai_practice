## Section A — Short Theory Questions

### 1. What is the difference between Matplotlib and Seaborn? When would you prefer one over the other?

**Matplotlib** is a low-level plotting library that gives full control over every element of a chart, but requires more code to make plots look polished. **Seaborn** is built on top of Matplotlib and provides high-level functions for creating attractive statistical plots with much less code, along with built-in themes and color palettes.

Matplotlib is preferred when we need fine-grained customization or are building a plot type Seaborn doesn't support. Seaborn is preferred for quick, visually appealing statistical visualizations, especially when working directly with Pandas DataFrames.

---

### 2. What is the difference between a univariate and a bivariate plot? Give one example of each.

A **univariate plot** displays and analyzes a single variable at a time, showing its distribution. Example: a histogram of students' marks.

A **bivariate plot** displays the relationship between two variables. Example: a scatter plot of study hours vs. marks.

---

### 3. What does a boxplot show, and how does it help detect outliers?

A **boxplot** displays the distribution of data through five key statistics: minimum, first quartile (Q1), median, third quartile (Q3), and maximum, along with "whiskers" that show the normal data range.

Any data points that fall outside the whiskers (beyond 1.5 times the interquartile range) are plotted as individual points and are considered **outliers** — making them easy to spot visually.

---

### 4. What is the purpose of a heatmap, and what kind of data is it best suited for?

A **heatmap** uses color intensity to represent the values in a matrix, making it easy to visually identify patterns, correlations, or concentrations of values at a glance.

It is best suited for **numerical, matrix-like data**, such as correlation matrices between variables, where color helps quickly show which variables are strongly or weakly related.

## Section B - Setup: Creating Sample Student Performance Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "Student": ["Ali", "Sara", "Ahmed", "Ayesha", "Bilal"] * 3,
    "Subject": ["Math"]*5 + ["Physics"]*5 + ["Chemistry"]*5,
    "Marks": [78, 88, 65, 92, 70, 82, 91, 60, 85, 75, 88, 79, 70, 95, 68],
    "Attendance": [90, 95, 80, 98, 85, 88, 92, 75, 96, 82, 91, 89, 78, 99, 81],
    "StudyHours": [5, 6, 3, 7, 4, 5, 6, 2, 7, 4, 6, 5, 3, 8, 4],
    "Gender": ["M", "F", "M", "F", "M"] * 3
})

print(df)
```

## Section B - Task 1: Line Plot - Marks Trend Across 5 Tests

```python
tests = ["Test1", "Test2", "Test3", "Test4", "Test5"]
ali_marks = [70, 75, 68, 82, 90]

plt.plot(tests, ali_marks, marker='o')
plt.title("Ali's Marks Trend Across 5 Tests")
plt.xlabel("Test")
plt.ylabel("Marks")
plt.show()
```

## Section B - Task 2: Bar Chart - Average Marks of 5 Students Across 3 Subjects

```python
avg_marks = df.groupby("Student")["Marks"].mean()

plt.bar(avg_marks.index, avg_marks.values, color='skyblue')
plt.title("Average Marks per Student")
plt.xlabel("Student")
plt.ylabel("Average Marks")
plt.show()
```

## Section B - Task 3: Distribution Plot - Spread of Marks Across the Class

```python
sns.histplot(df["Marks"], kde=True, color='purple')
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

## Section B - Task 4: Boxplot - Marks Distribution Across Subjects

```python
sns.boxplot(x="Subject", y="Marks", data=df)
plt.title("Marks Distribution by Subject")
plt.show()
```

## Section B - Task 5: Violinplot - Marks Distribution by Gender

```python
sns.violinplot(x="Gender", y="Marks", data=df)
plt.title("Marks Distribution by Gender")
plt.show()
```

## Section B - Task 6: Pairplot - Study Hours, Attendance, Marks

```python
sns.pairplot(df[["StudyHours", "Attendance", "Marks"]])
plt.show()
```

## Section B - Task 7: Heatmap - Correlation Matrix

```python
corr = df[["Marks", "Attendance", "StudyHours"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

## Section B - Task 8: Customized Plot - Boxplot with Full Styling

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="Subject", y="Marks", data=df, hue="Subject", palette="Set2", legend=False)
plt.title("Marks Distribution by Subject (Customized)", fontsize=14, fontweight='bold')
plt.xlabel("Subject", fontsize=12)
plt.ylabel("Marks", fontsize=12)
plt.show()
```
