## Section A — Short Theory Questions

### 1. Differentiate between quantitative and qualitative data with examples.

**Quantitative data** is numerical data that can be measured and counted. Example: height, weight, marks obtained, age.

**Qualitative data** is descriptive, non-numerical data that represents categories or characteristics. Example: gender, color, city name, blood group.

---

### 2. What is the difference between mean, median, and mode? When would median be preferred over mean?

- **Mean**: The average of all values (sum of values divided by count).
- **Median**: The middle value when data is arranged in order.
- **Mode**: The most frequently occurring value in the dataset.

**Median is preferred over mean** when the dataset contains outliers or is skewed, since extreme values heavily affect the mean but have little effect on the median.

---

### 3. What do variance and standard deviation measure, and how are they related?

**Variance** measures how spread out the data values are from the mean (average of squared differences from the mean).

**Standard deviation** is the square root of variance, and it expresses the spread in the same unit as the original data, making it easier to interpret.

They are related as: **Standard Deviation = √Variance**

---

### 4. Explain Z-score and how it is used to detect outliers.

A **Z-score** tells us how many standard deviations a data point is away from the mean. It is calculated as:

Z = (x - mean) / standard deviation

A value is typically considered an **outlier** if its Z-score is greater than +3 or less than -3, meaning it lies far away from the rest of the data.

## Section B - Task 1: Mean, Median, Mode

```python
marks = [56, 78, 90, 45, 78, 88, 67, 92, 78, 60]

n = len(marks)
mean_manual = sum(marks) / n

sorted_marks = sorted(marks)
mid = n // 2
median_manual = (sorted_marks[mid-1] + sorted_marks[mid]) / 2 if n % 2 == 0 else sorted_marks[mid]

mode_manual = max(set(marks), key=marks.count)

import statistics as stats
mean_builtin = stats.mean(marks)
median_builtin = stats.median(marks)
mode_builtin = stats.mode(marks)

print("Manual -> Mean:", mean_manual, "Median:", median_manual, "Mode:", mode_manual)
print("Built-in -> Mean:", mean_builtin, "Median:", median_builtin, "Mode:", mode_builtin)
```

## Section B - Task 2: Variance, Standard Deviation, Coefficient of Variation

```python
variance = stats.variance(marks)
std_dev = stats.stdev(marks)
cv = (std_dev / mean_builtin) * 100

print("Variance:", variance)
print("Standard Deviation:", std_dev)
print("Coefficient of Variation:", cv, "%")
```

## Section B - Task 3: Z-Scores and Outlier Detection

```python
import numpy as np

marks_arr = np.array(marks)
z_scores = (marks_arr - mean_builtin) / std_dev

outliers = marks_arr[np.abs(z_scores) > 3]

print("Z-scores:", z_scores)
print("Outliers:", outliers)
```

## Section B - Task 4: Correlation Coefficient - Study Hours vs Marks

```python
study_hours = [2, 4, 5, 1, 4, 5, 3, 6, 4, 2]

correlation = np.corrcoef(study_hours, marks)[0, 1]
print("Correlation coefficient:", correlation)
```

## Section 2 - Task 1: NumPy Arrays - Creation, Reshape, Indexing, Slicing

```python
arr_from_list = np.array([1, 2, 3, 4, 5, 6])
arr_arange = np.arange(1, 13)
arr_random = np.random.randint(1, 100, 6)

reshaped = arr_arange.reshape(3, 4)

print("From list:", arr_from_list)
print("Arange:", arr_arange)
print("Random:", arr_random)
print("Reshaped:\n", reshaped)

print("Indexing element:", reshaped[1, 2])
print("Slicing row:", reshaped[0, :])
print("Slicing column:", reshaped[:, 1])
```

## Section 2 - Task 3: Pandas DataFrame - Basic Info

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Bilal"],
    "Subject": ["Math", "Physics", "Math", "Chemistry", "Physics"],
    "Marks": [85, 90, 78, 92, 60]
})

print(df.head())
print("Shape:", df.shape)
print("Dtypes:\n", df.dtypes)
```

## Section 2 - Task 4: Data Selection and Filtering

```python
print(df[df["Marks"] > 80])
print(df.loc[0:2, ["Name", "Marks"]])
print(df.iloc[1:4, 0:2])
```

## Section 2 - Task 5: Handling Missing Data

```python
df_missing = df.copy()
df_missing.loc[2, "Marks"] = None
df_missing.loc[4, "Subject"] = None

print("Null values:\n", df_missing.isnull().sum())

df_filled = df_missing.fillna({"Marks": df_missing["Marks"].mean(), "Subject": "Unknown"})
print(df_filled)

df_dropped = df_missing.dropna()
print(df_dropped)
```

## Section 2 - Task 6: Groupby - Average Marks per Subject

```python
avg_marks = df.groupby("Subject")["Marks"].mean()
print(avg_marks)
```

## Section 2 - Task 7: Merge and Concatenate DataFrames

```python
df1 = pd.DataFrame({"Name": ["Ali", "Sara"], "Marks": [85, 90]})
df2 = pd.DataFrame({"Name": ["Ahmed", "Ayesha"], "Marks": [78, 92]})

concatenated = pd.concat([df1, df2], ignore_index=True)
print("Concatenated:\n", concatenated)

df3 = pd.DataFrame({"Name": ["Ali", "Sara"], "City": ["Lahore", "Karachi"]})
merged = pd.merge(df1, df3, on="Name")
print("Merged:\n", merged)
```
