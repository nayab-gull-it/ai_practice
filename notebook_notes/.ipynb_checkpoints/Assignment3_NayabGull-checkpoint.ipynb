## Section A — Short Theory Questions

### 1. Differentiate between a class and an object with a real-world example.

A **class** is a blueprint or template that defines the properties (attributes) and behaviors (methods) that its objects will have. An **object** is a real instance created from that class, with actual values.

**Example:** `Car` can be a class defining attributes like `brand`, `color`, `speed`. `my_car = Car("Toyota", "White", 120)` is an object — an actual car built from that blueprint.

---

### 2. What is the difference between instance variables and class variables?

- **Instance variables**: Belong to a specific object. Each object has its own copy, defined usually inside `__init__` using `self`. Example: `self.name`.
- **Class variables**: Shared by all objects of the class. Defined directly inside the class body (outside methods), and changing it affects all instances unless overridden individually.

---

### 3. Explain the difference between Single, Multilevel, Multiple, and Hierarchical inheritance.

- **Single inheritance**: One child class inherits from one parent class.
- **Multilevel inheritance**: A chain of inheritance — a class inherits from a class, which itself inherits from another class (grandparent → parent → child).
- **Multiple inheritance**: A single child class inherits from more than one parent class at the same time.
- **Hierarchical inheritance**: Multiple child classes inherit from the same single parent class.

---

### 4. What is encapsulation, and how is it implemented in Python using access specifiers?

**Encapsulation** is the OOP principle of bundling data (attributes) and methods together inside a class, while restricting direct access to some of that data to protect it from unintended modification.

In Python, it is implemented using naming conventions:
- **Public** (`name`): Accessible from anywhere.
- **Protected** (`_name`): Meant for internal use within the class and its subclasses (convention only, not strictly enforced).
- **Private** (`__name`): Name-mangled so it cannot be accessed directly from outside the class, enforcing stronger restriction.

---

### 5. What is the difference between method overriding and operator overloading?

- **Method overriding**: A subclass provides its own implementation of a method that is already defined in its parent class, changing its behavior for the subclass.
- **Operator overloading**: Redefining how a built-in operator (like `+`, `-`, `==`) behaves for objects of a custom class, usually by implementing special/dunder methods like `__add__()`.

## Task 1: Perfume class with instance variables + display method

```python
class Perfume:
    def __init__(self, name, brand, price, volume_ml):
        self.name = name
        self.brand = brand
        self.price = price
        self.volume_ml = volume_ml

    def display(self):
        print(f"{self.name} by {self.brand} - Rs.{self.price} - {self.volume_ml}ml")

p1 = Perfume("Aroma Bloom", "Aroma House", 2500, 100)
p1.display()
```

## Task 2: class variable total_perfumes

```python
class Perfume:
    total_perfumes = 0

    def __init__(self, name, brand, price, volume_ml):
        self.name = name
        self.brand = brand
        self.price = price
        self.volume_ml = volume_ml
        Perfume.total_perfumes += 1

    def display(self):
        print(f"{self.name} by {self.brand} - Rs.{self.price} - {self.volume_ml}ml")

p1 = Perfume("Aroma Bloom", "Aroma House", 2500, 100)
p2 = Perfume("Musk Royale", "Aroma House", 3200, 50)
print("Total perfumes created:", Perfume.total_perfumes)
```

## Task 3: Product base class → Perfume and Accessory subclasses

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def display(self):
        print(f"{self.name} - Rs.{self.price}")

class Perfume(Product):
    def __init__(self, name, price, brand, volume_ml):
        super().__init__(name, price)
        self.brand = brand
        self.volume_ml = volume_ml

class Accessory(Product):
    def __init__(self, name, price, accessory_type):
        super().__init__(name, price)
        self.accessory_type = accessory_type

p = Perfume("Aroma Bloom", 2500, "Aroma House", 100)
a = Accessory("Gift Box", 300, "Packaging")
p.display()
a.display()
```

## Task 4: Multiple inheritance — Discountable + Perfume → PremiumPerfume

```python
class Discountable:
    def apply_discount(self, price, percent=10):
        return price - (price * percent / 100)

class PremiumPerfume(Discountable, Perfume):
    pass

pp = PremiumPerfume("Oud Elite", 5000, "Aroma House", 75)
print("Discounted price:", pp.apply_discount(pp.price))
print(PremiumPerfume.__mro__)
```

## Task 6: overload + operator → ComboPack

```python
class ComboPack:
    def __init__(self, description, total_price):
        self.description = description
        self.total_price = total_price

    def display(self):
        print(f"{self.description} - Rs.{self.total_price}")

class Perfume:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __add__(self, other):
        return ComboPack(f"{self.name} + {other.name}", self.price + other.price)

p1 = Perfume("Aroma Bloom", 2500)
p2 = Perfume("Musk Royale", 3200)
combo = p1 + p2
combo.display()
```

## Task 7: abstract class Product with calculate_final_price()

```python
from abc import ABC, abstractmethod

class Product(ABC):
    def __init__(self, name, price):
        self.name = name
        self.price = price

    @abstractmethod
    def calculate_final_price(self):
        pass

class Perfume(Product):
    def calculate_final_price(self):
        return self.price + (self.price * 0.05)

class GiftSet(Product):
    def calculate_final_price(self):
        return self.price - (self.price * 0.15)

p = Perfume("Aroma Bloom", 2500)
g = GiftSet("Festive Gift Set", 4000)
print(p.calculate_final_price())
print(g.calculate_final_price())
```

## Task 8: dynamic polymorphism with Product-type reference

```python
items = [Perfume("Aroma Bloom", 2500), GiftSet("Festive Gift Set", 4000)]
for item in items:
    ref: Product = item
    print(f"{ref.name} final price: {ref.calculate_final_price()}")
```
