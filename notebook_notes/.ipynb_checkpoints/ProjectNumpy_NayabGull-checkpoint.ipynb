# Project - NumPy

## Task 1: generate Dataset (1,000 random integers, 1 to 10,000)

```python
import numpy as np

data = np.random.randint(1, 10001, 1000)
print("Dataset generated with shape:", data.shape)
print(data)
```

## Task 2: 10 Largest Values in the Dataset

```python
largest_10 = np.sort(data)[-10:][::-1]
print("10 Largest values:", largest_10)
```

## Task 3: 10 Smallest Values in the Dataset

```python
smallest_10 = np.sort(data)[:10]
print("10 Smallest values:", smallest_10)
```

## Task 4: Unique Values in the Dataset

```python
unique_values = np.unique(data)
print("Number of unique values:", len(unique_values))
print(unique_values)
```

## Task 5: Frequency of Each Unique Value

```python
values, counts = np.unique(data, return_counts=True)
frequency = dict(zip(values, counts))
print(frequency)
```

## Task 6: Count of Even and Odd Numbers

```python
even_count = np.sum(data % 2 == 0)
odd_count = np.sum(data % 2 != 0)
print("Even numbers:", even_count)
print("Odd numbers:", odd_count)
```

## Task 7: Numbers Divisible by 7

```python
divisible_by_7 = data[data % 7 == 0]
print("Count divisible by 7:", len(divisible_by_7))
print(divisible_by_7)
```

## Task 8: Percentage of Values Greater than 5000

```python
percentage_gt_5000 = (np.sum(data > 5000) / len(data)) * 100
print("Percentage greater than 5000:", percentage_gt_5000, "%")
```

## Task 9: Dataset Sorted in Ascending Order

```python
ascending = np.sort(data)
print(ascending)
```

## Task 10: Dataset Sorted in Descending Order

```python
descending = np.sort(data)[::-1]
print(descending)
```

## Task 11: Index (Position) of Maximum Value

```python
max_index = np.argmax(data)
print("Index of max value:", max_index, "-> Value:", data[max_index])
```

## Task 12: Index (Position) of Minimum Value

```python
min_index = np.argmin(data)
print("Index of min value:", min_index, "-> Value:", data[min_index])
```

## Conclusion

This assignment helped me understand how powerful and efficient NumPy is for handling large datasets compared to standard Python lists. Using NumPy, I was able to generate 1,000 random integers and perform operations like sorting, finding unique values, calculating frequencies, and locating maximum/minimum values in just a single line of code for each task.

Built-in functions such as `np.sort()`, `np.unique()`, `np.argmax()`, and `np.argmin()` eliminated the need for writing manual loops, making the code shorter, cleaner, and much faster to execute. Vectorized operations like `data % 2 == 0` allowed conditions to be applied to the entire array at once, instead of looping through each element individually as we would with a normal Python list.

Overall, NumPy proved to be extremely useful for real-world data analysis tasks where large amounts of numerical data need to be processed quickly and efficiently. Key takeaways include the importance of vectorization, the ease of statistical analysis, and how much simpler NumPy makes numerical computations compared to using loops and manual operations on Python lists.m
