# 🌍 The 17 Sustainable Development Goals (SDGs)

This notebook covers all **17 Sustainable Development Goals (SDGs)** adopted by the United Nations in 2015 as part of the 2030 Agenda for Sustainable Development.

Each goal below includes:
- Title and official icon
- Key focus points (sub-headings)
- A short description for each point


## SDG 1: No Poverty

![SDG 1 - No Poverty](https://open-sdg.github.io/sdg-translations/assets/img/goals/en/1.png)


### Ending Extreme Poverty

The goal aims to eliminate extreme poverty for all people everywhere, currently measured as living on less than $1.90 a day, by ensuring access to basic resources and opportunities.

### Social Protection Systems

It calls for implementing nationally appropriate social protection systems, including floors, to cover the poor and vulnerable, reducing exposure to economic shocks.

### Equal Rights to Resources

Ensuring that all men and women, particularly the poor, have equal rights to economic resources, land ownership, inheritance, and access to financial services.

## SDG 2: Zero Hunger

![SDG 2 - Zero Hunger](https://open-sdg.github.io/sdg-translations/assets/img/goals/en/2.png)


### Ending Hunger and Malnutrition

This goal focuses on ending hunger and ensuring access for all people, especially children, to safe, nutritious, and sufficient food throughout the year.

### Sustainable Agriculture

It promotes sustainable food production systems and resilient agricultural practices that increase productivity without degrading land and ecosystems.

### Supporting Small-Scale Farmers

Special attention is given to doubling the income and productivity of small-scale food producers through fair access to land, markets, and financial services.

## SDG 3: Good Health and Well-being

![SDG 3 - Good Health and Well-being](https://open-sdg.github.io/sdg-translations/assets/img/goals/en/3.png)


### Reducing Maternal and Child Mortality

The goal targets a significant reduction in maternal deaths and preventable deaths of newborns and children under five years of age.

### Combating Diseases

It aims to end epidemics of AIDS, tuberculosis, malaria, and other communicable diseases, along with reducing deaths from non-communicable diseases.

### Universal Health Coverage

Achieving universal health coverage ensures access to quality essential healthcare services and affordable medicines for everyone.

## SDG 4: Quality Education

![SDG 4 - Quality Education](https://open-sdg.github.io/sdg-translations/assets/img/goals/en/4.png)


### Inclusive and Equitable Education

This goal seeks to ensure free, equitable, and quality primary and secondary education for all girls and boys.

### Access to Technical and Vocational Training

It promotes equal access to affordable technical, vocational, and higher education, including university, for both men and women.

### Literacy and Numeracy for All

The goal aims to ensure that all youth and a substantial proportion of adults achieve literacy and numeracy skills.

## SDG 5: Gender Equality

![SDG 5 - Gender Equality](https://open-sdg.github.io/sdg-translations/assets/img/goals/en/5.png)


### Ending Discrimination Against Women

The goal calls for ending all forms of discrimination against women and girls everywhere, recognizing it as a fundamental human right.

### Eliminating Violence and Harmful Practices

It seeks to eliminate all forms of violence against women, including trafficking, sexual exploitation, and harmful practices like child marriage.

### Women's Leadership and Participation

Ensuring women's full participation and equal opportunities for leadership in political, economic, and public life is a key focus.

## SDG 6: Clean Water and Sanitation

![SDG 6 - Clean Water and Sanitation](https://open-sdg.github.io/sdg-translations/assets/img/goals/en/6.png)


### Universal Access to Safe Water

The goal aims to achieve universal and equitable access to safe and affordable drinking water for all by 2030.

### Improved Sanitation and Hygiene

It targets access to adequate and equitable sanitation and hygiene, ending open defecation and protecting vulnerable groups.

### Protecting Water Ecosystems

This includes protecting and restoring water-related ecosystems such as rivers, lakes, and wetlands to ensure long-term water security.

## SDG 7: Affordable and Clean Energy

![SDG 7 - Affordable and Clean Energy](https://open-sdg.github.io/sdg-translations/assets/img/goals/en/7.png)


### Universal Access to Energy

This goal aims to ensure universal access to affordable, reliable, and modern energy services for everyone.

### Increasing Renewable Energy Share

It promotes a substantial increase in the share of renewable energy sources like solar and wind in the global energy mix.

### Improving Energy Efficiency

Doubling the global rate of improvement in energy efficiency is a key target to reduce waste and environmental impact.

## SDG 8: Decent Work and Economic Growth

![SDG 8 - Decent Work and Economic Growth](https://open-sdg.github.io/sdg-translations/assets/img/goals/en/8.png)


### Sustainable Economic Growth

The goal promotes sustained, inclusive, and sustainable economic growth, aiming for at least 7% GDP growth per year in least developed countries.

### Full and Productive Employment

It calls for full and productive employment and decent work for all women and men, including equal pay for equal work.

### Ending Forced Labor and Child Labor

This includes taking immediate measures to eradicate forced labor, modern slavery, human trafficking, and child labor in all its forms.

## SDG 9: Industry, Innovation and Infrastructure

![SDG 9 - Industry, Innovation and Infrastructure](https://open-sdg.github.io/sdg-translations/assets/img/goals/en/9.png)


### Resilient Infrastructure

The goal focuses on developing quality, reliable, and resilient infrastructure to support economic development and human well-being.

### Inclusive and Sustainable Industrialization

It promotes inclusive and sustainable industrialization, significantly raising industry's share of employment and GDP.

### Fostering Innovation

Enhancing scientific research and upgrading technological capabilities encourages innovation and increases the number of R&D workers.

## SDG 10: Reduced Inequalities

![SDG 10 - Reduced Inequalities](https://open-sdg.github.io/sdg-translations/assets/img/goals/en/10.png)


### Reducing Income Inequality

This goal aims to progressively achieve and sustain income growth of the bottom 40% of the population at a rate higher than the national average.

### Social, Economic, and Political Inclusion

It promotes the social, economic, and political inclusion of all people, irrespective of age, sex, disability, race, or origin.

### Fair Migration Policies

Facilitating safe, orderly, and responsible migration through well-managed policies is another key focus area.

## SDG 11: Sustainable Cities and Communities

![SDG 11 - Sustainable Cities and Communities](https://open-sdg.github.io/sdg-translations/assets/img/goals/en/11.png)


### Access to Safe Housing

The goal aims to ensure access for all to adequate, safe, and affordable housing and basic services, and to upgrade slums.

### Sustainable Transport Systems

It promotes access to safe, affordable, and sustainable transport systems, improving road safety for all.

### Reducing Environmental Impact of Cities

This includes reducing the adverse environmental impact of cities, particularly air quality and waste management.

## SDG 12: Responsible Consumption and Production

![SDG 12 - Responsible Consumption and Production](https://open-sdg.github.io/sdg-translations/assets/img/goals/en/12.png)


### Sustainable Management of Resources

The goal promotes the sustainable management and efficient use of natural resources across production and consumption cycles.

### Reducing Food Waste

It targets halving global food waste per capita at the retail and consumer levels, along with reducing food losses in production.

### Responsible Waste Management

Achieving environmentally sound management of chemicals and waste throughout their life cycle is a core target.

## SDG 13: Climate Action

![SDG 13 - Climate Action](https://open-sdg.github.io/sdg-translations/assets/img/goals/en/13.png)


### Strengthening Climate Resilience

This goal focuses on strengthening resilience and adaptive capacity to climate-related hazards and natural disasters.

### Integrating Climate Policy

It calls for integrating climate change measures into national policies, strategies, and planning.

### Climate Education and Awareness

Improving education, awareness-raising, and human capacity on climate change mitigation and adaptation is emphasized.

## SDG 14: Life Below Water

![SDG 14 - Life Below Water](https://open-sdg.github.io/sdg-translations/assets/img/goals/en/14.png)


### Reducing Marine Pollution

The goal aims to prevent and significantly reduce marine pollution of all kinds, particularly from land-based activities.

### Sustainable Fishing Practices

It promotes ending overfishing and destructive fishing practices to restore fish stocks to sustainable levels.

### Conserving Coastal Ecosystems

Sustainably managing and protecting marine and coastal ecosystems helps avoid significant adverse impacts on ocean life.

## SDG 15: Life on Land

![SDG 15 - Life on Land](https://open-sdg.github.io/sdg-translations/assets/img/goals/en/15.png)


### Conserving Terrestrial Ecosystems

This goal focuses on conserving, restoring, and sustainably using terrestrial ecosystems such as forests, wetlands, and mountains.

### Combating Desertification

It aims to combat desertification and restore degraded land and soil affected by drought and flooding.

### Protecting Biodiversity

Urgent action to reduce the degradation of natural habitats and halt biodiversity loss is a key priority.

## SDG 16: Peace, Justice and Strong Institutions

![SDG 16 - Peace, Justice and Strong Institutions](https://open-sdg.github.io/sdg-translations/assets/img/goals/en/16.png)


### Reducing Violence

The goal aims to significantly reduce all forms of violence and related death rates everywhere.

### Promoting Rule of Law

It promotes the rule of law at national and international levels, ensuring equal access to justice for all.

### Building Effective Institutions

Developing effective, accountable, and transparent institutions at all levels is essential for good governance.

## SDG 17: Partnerships for the Goals

![SDG 17 - Partnerships for the Goals](https://open-sdg.github.io/sdg-translations/assets/img/goals/en/17.png)


### Strengthening Global Partnerships

This goal emphasizes strengthening the means of implementation and revitalizing global partnerships for sustainable development.

### Mobilizing Financial Resources

It calls for mobilizing additional financial resources for developing countries from multiple sources, including international support.

### Enhancing Technology and Capacity Building

Promoting the development, transfer, and dissemination of environmentally sound technologies is key to supporting developing nations.
