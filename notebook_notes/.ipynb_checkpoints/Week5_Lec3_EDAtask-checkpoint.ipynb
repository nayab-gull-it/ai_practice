```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
# Create an array of 20 random numbers

numbers = np.random.randint(1, 101, 20)

print("Random Numbers:")
print(numbers)
```

```python
# Mean, Maximum and Minimum

print("Mean :", np.mean(numbers))
print("Maximum :", np.max(numbers))
print("Minimum :", np.min(numbers))
```

```python
student_data = {
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha", "Hamza"],
    "Subject": ["Math", "Physics", "Chemistry", "Biology", "English"],
    "Marks": [85, 92, 78, 95, 88]
}

df = pd.DataFrame(student_data)

df
```

```python
top_student = df.loc[df["Marks"].idxmax()]

print("Student with Highest Marks")
top_student
```

```python
plt.figure(figsize=(7,5))

plt.bar(df["Name"], df["Marks"], color="skyblue")

plt.title("Students Marks")
plt.xlabel("Student Name")
plt.ylabel("Marks")

plt.show()
```

```python
iris = sns.load_dataset("iris")

iris.head()
```

```python
correlation = iris.select_dtypes(include="number").corr()

correlation
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    linewidths=1
)

plt.title("Correlation Heatmap of Iris Dataset")

plt.show()
```
