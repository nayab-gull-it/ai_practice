## ⚙️ Setup — Import All Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statistics as stats
from scipy import stats as sp
from scipy.stats import binom, poisson, norm

sns.set_theme(style="whitegrid")
print("All libraries loaded ✅")
```

## 📋 Sample Data — Student Marks

```python
marks        = [55, 70, 85, 60, 90, 75, 80, 65, 70, 95, 50, 72, 68, 88, 76]
study_hours  = [2,  3,  5,  1,  6,  4,  7,  3,  5,  8,  1,  4,  3,  6,  5]
print("Marks:", marks)
```

---
# 📊 PART 1 — Types of Data

| Type | Sub-type | Example |
|---|---|---|
| **Quantitative** | Continuous | Height, Weight |
| **Quantitative** | Discrete | No. of students |
| **Qualitative** | Nominal | Colors, Gender |
| **Qualitative** | Ordinal | Grades A, B, C |
| **Qualitative** | Binary | Yes/No |

### Visualizing Data Types — Bar (Nominal) vs Histogram (Continuous)

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Nominal (Qualitative)
colors  = ['Red', 'Blue', 'Green', 'Yellow']
counts  = [30, 45, 20, 15]
axes[0].bar(colors, counts, color=['red','blue','green','gold'], edgecolor='black')
axes[0].set_title("Nominal Data — Favorite Colors")
axes[0].set_ylabel("Count")

# Continuous (Quantitative)
heights = np.random.normal(170, 8, 200)
axes[1].hist(heights, bins=20, color='steelblue', edgecolor='black')
axes[1].set_title("Continuous Data — Student Heights")
axes[1].set_xlabel("Height (cm)")

plt.tight_layout()
plt.show()
```

---
# 📍 PART 2 — Measures of Central Tendency

| Measure | What it finds |
|---|---|
| **Mean** | Average |
| **Median** | Middle value |
| **Mode** | Most repeated |

## 1️⃣ Mean, Median, Mode — Calculate

```python
mean   = np.mean(marks)
median = np.median(marks)
mode   = stats.mode(marks)

print(f"Mean   = {mean:.2f}")
print(f"Median = {median}")
print(f"Mode   = {mode}")
```

### 📉 Graph — Mean, Median, Mode on Histogram

```python
plt.figure(figsize=(9, 4))
plt.hist(marks, bins=8, color='lightsteelblue', edgecolor='black', alpha=0.8)
plt.axvline(mean,   color='red',    linestyle='--', linewidth=2, label=f'Mean = {mean:.1f}')
plt.axvline(median, color='green',  linestyle='-',  linewidth=2, label=f'Median = {median}')
plt.axvline(mode,   color='orange', linestyle=':',  linewidth=2, label=f'Mode = {mode}')
plt.title("Mean, Median & Mode on Histogram")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

### Effect of Outlier on Mean vs Median

```python
salaries        = [30, 32, 31, 29, 33, 500]   # 500K is outlier
mean_sal        = np.mean(salaries)
median_sal      = np.median(salaries)

plt.figure(figsize=(9, 4))
plt.plot(salaries, 'bo-', markersize=8, label='Salaries')
plt.axhline(mean_sal,   color='red',   linestyle='--', label=f'Mean = {mean_sal:.0f}K')
plt.axhline(median_sal, color='green', linestyle='-',  label=f'Median = {median_sal}K')
plt.title("Outlier Effect: Mean vs Median")
plt.ylabel("Salary (K)")
plt.legend()
plt.show()
print("Median is more reliable with outliers ✅")
```

# 📏 PART 3 — Measures of Dispersion

| Measure | Meaning |
|---|---|
| **Range** | Max - Min |
| **Variance** | Avg squared distance from mean |
| **Std Dev** | √Variance (same unit as data) |
| **CV** | Std/Mean × 100% |

## 2️⃣ Variance & Standard Deviation

```python
std = np.std(marks, ddof=1)
var = np.var(marks, ddof=1)
cv  = (std / np.mean(marks)) * 100

print(f"Range  = {max(marks) - min(marks)}")
print(f"Std Dev= {std:.2f}")
print(f"Var    = {var:.2f}")
print(f"CV     = {cv:.2f}%")
```

### 📉 Graph — Std Dev Bands on Data

```python
mean = np.mean(marks)
std  = np.std(marks, ddof=1)

plt.figure(figsize=(9, 4))
plt.plot(marks, 'o', color='steelblue', markersize=8, label='Marks')
plt.axhline(mean,        color='red',   linewidth=2,   label=f'Mean = {mean:.1f}')
plt.axhline(mean + std,  color='green', linestyle='--', label=f'+1 Std = {mean+std:.1f}')
plt.axhline(mean - std,  color='green', linestyle='--', label=f'-1 Std = {mean-std:.1f}')
plt.fill_between(range(len(marks)), mean - std, mean + std, alpha=0.1, color='green')
plt.title("Marks with Standard Deviation Bands")
plt.ylabel("Marks")
plt.xlabel("Student")
plt.legend()
plt.show()
```

### Comparing Two Classes — High vs Low Spread

```python
class_a = np.random.normal(70, 3,  100)  # low spread
class_b = np.random.normal(70, 15, 100)  # high spread

plt.figure(figsize=(9, 4))
plt.hist(class_a, bins=20, alpha=0.6, color='steelblue', label=f'Class A  Std={np.std(class_a):.1f}')
plt.hist(class_b, bins=20, alpha=0.6, color='tomato',    label=f'Class B  Std={np.std(class_b):.1f}')
plt.title("Low Spread vs High Spread")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

## 3️⃣ Skewness — Is Data Tilted?

| Value | Shape |
|---|---|
| `= 0` | Symmetric |
| `> 0` | Right skewed (tail right) |
| `< 0` | Left skewed (tail left) |

```python
right_skew = np.random.exponential(scale=2, size=500)
symmetric  = np.random.normal(0, 1, 500)
left_skew  = -np.random.exponential(scale=2, size=500)

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].hist(left_skew,  bins=30, color='tomato',      edgecolor='black')
axes[0].set_title(f"Left Skewed\nskew={sp.skew(left_skew):.2f}")

axes[1].hist(symmetric,  bins=30, color='steelblue',   edgecolor='black')
axes[1].set_title(f"Symmetric\nskew={sp.skew(symmetric):.2f}")

axes[2].hist(right_skew, bins=30, color='mediumseagreen', edgecolor='black')
axes[2].set_title(f"Right Skewed\nskew={sp.skew(right_skew):.2f}")

plt.tight_layout()
plt.show()
```

## 4️⃣ Kurtosis — Peak Shape

| Value | Shape | Name |
|---|---|---|
| `= 0` | Normal peak | Mesokurtic |
| `> 0` | Sharp peak | Leptokurtic |
| `< 0` | Flat peak | Platykurtic |

```python
normal  = np.random.normal(0, 1, 1000)
sharp   = np.random.laplace(0, 1, 1000)  # sharp peak
flat    = np.random.uniform(-3, 3, 1000) # flat

plt.figure(figsize=(9, 4))
sns.kdeplot(normal, label=f'Normal  kurt={sp.kurtosis(normal):.2f}',  color='steelblue', linewidth=2)
sns.kdeplot(sharp,  label=f'Sharp   kurt={sp.kurtosis(sharp):.2f}',   color='tomato',    linewidth=2)
sns.kdeplot(flat,   label=f'Flat    kurt={sp.kurtosis(flat):.2f}',    color='green',     linewidth=2)
plt.title("Kurtosis Comparison")
plt.legend()
plt.show()
```

# 📌 PART 4 — Measures of Position

| Measure | What it does |
|---|---|
| **Z-Score** | Distance from mean in std units |
| **Percentile** | % of data below a value |
| **Quartile** | Splits data into 4 parts |

## 5️⃣ Z-Score — Formula: Z = (x - mean) / std

```python
mean     = np.mean(marks)
std      = np.std(marks, ddof=1)
z_scores = [(x - mean) / std for x in marks]

for m, z in zip(marks[:6], z_scores[:6]):
    print(f"Mark={m}  Z={z:+.2f}")
```

### 📉 Graph — Z-Scores for Each Student

```python
colors = ['tomato' if z < 0 else 'steelblue' for z in z_scores]

plt.figure(figsize=(10, 4))
plt.bar(range(len(z_scores)), z_scores, color=colors, edgecolor='black')
plt.axhline(0, color='black', linewidth=1.5)
plt.axhline( 1, color='green',  linestyle='--', label='+1 std')
plt.axhline(-1, color='green',  linestyle='--', label='-1 std')
plt.title("Z-Scores of Students (Blue=Above Mean, Red=Below)")
plt.xlabel("Student")
plt.ylabel("Z-Score")
plt.legend()
plt.show()
```

## 6️⃣ Percentiles

```python
p10 = np.percentile(marks, 10)
p25 = np.percentile(marks, 25)
p50 = np.percentile(marks, 50)
p75 = np.percentile(marks, 75)
p90 = np.percentile(marks, 90)

print(f"P10={p10}  P25={p25}  P50={p50}  P75={p75}  P90={p90}")
```

### 📉 Graph — Percentile Lines on Histogram

```python
plt.figure(figsize=(9, 4))
plt.hist(marks, bins=8, color='lightsteelblue', edgecolor='black')
plt.axvline(p25, color='blue',   linestyle='--', label=f'P25 = {p25}')
plt.axvline(p50, color='green',  linestyle='-',  label=f'P50 = {p50}')
plt.axvline(p75, color='orange', linestyle='--', label=f'P75 = {p75}')
plt.axvline(p90, color='red',    linestyle=':',  label=f'P90 = {p90}')
plt.title("Percentile Lines on Histogram")
plt.xlabel("Marks")
plt.legend()
plt.show()
```

## 7️⃣ Quartiles & IQR — Box Plot

```python
Q1  = np.percentile(marks, 25)
Q2  = np.percentile(marks, 50)
Q3  = np.percentile(marks, 75)
IQR = Q3 - Q1

print(f"Q1={Q1}  Q2={Q2}  Q3={Q3}  IQR={IQR}")
print(f"Outlier if < {Q1 - 1.5*IQR} or > {Q3 + 1.5*IQR}")
```

### 📉 Graph — Box Plot showing Q1, Q2, Q3

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Seaborn box plot
sns.boxplot(x=marks, ax=axes[0], color='lightgreen')
axes[0].set_title("Box Plot — Student Marks")
axes[0].set_xlabel("Marks")

# With outlier
marks_with_outlier = marks + [20, 120]
sns.boxplot(x=marks_with_outlier, ax=axes[1], color='lightsalmon')
axes[1].set_title("Box Plot — With Outliers")
axes[1].set_xlabel("Marks")

plt.tight_layout()
plt.show()
```

# 🔗 PART 5 — Correlation

| Value | Relationship |
|---|---|
| `+1` | Perfect positive |
| `0` | No relation |
| `-1` | Perfect negative |

## 8️⃣ Correlation Coefficient

```python
corr = np.corrcoef(study_hours, marks)[0, 1]
print(f"Correlation (Hours vs Marks) = {corr:.4f}")
```

### 📉 Graph — Three Types of Correlation

```python
x = np.linspace(0, 10, 50)

pos = x + np.random.normal(0, 1, 50)
neg = -x + np.random.normal(0, 1, 50)
no  = np.random.normal(0, 3, 50)

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].scatter(x, pos, color='steelblue')
axes[0].set_title(f"Positive Correlation\nr = {np.corrcoef(x,pos)[0,1]:.2f}")

axes[1].scatter(x, no,  color='gray')
axes[1].set_title(f"No Correlation\nr = {np.corrcoef(x,no)[0,1]:.2f}")

axes[2].scatter(x, neg, color='tomato')
axes[2].set_title(f"Negative Correlation\nr = {np.corrcoef(x,neg)[0,1]:.2f}")

plt.tight_layout()
plt.show()
```

### 📉 Graph — Correlation Heatmap (Seaborn)

```python
import pandas as pd

df = pd.DataFrame({
    'Marks':       marks,
    'Study_Hours': study_hours,
    'Sleep_Hours': [7,6,8,5,9,7,8,6,7,9,5,7,6,8,7]
})

plt.figure(figsize=(6, 4))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Heatmap")
plt.show()
```

# 🎲 PART 7 — Probability

**Formula:** P(A) = Favorable ÷ Total outcomes

- 0 = impossible &nbsp; | &nbsp; 1 = certain

| Type | Formula |
|---|---|
| **Joint** | P(A∩B) = P(A) × P(B) |
| **Marginal** | P(A) alone |
| **Conditional** | P(A\|B) = P(A∩B) / P(B) |

## 1️⃣3️⃣ Basic Probability — Simulating Coin Toss

```python
tosses = np.random.choice(['Head', 'Tail'], size=1000)
heads  = np.sum(tosses == 'Head')

plt.figure(figsize=(6, 4))
plt.bar(['Head', 'Tail'],
        [np.sum(tosses=='Head'), np.sum(tosses=='Tail')],
        color=['steelblue', 'tomato'], edgecolor='black')
plt.title("1000 Coin Tosses — Probability Simulation")
plt.ylabel("Count")
plt.axhline(500, color='black', linestyle='--', label='Expected (500)')
plt.legend()
plt.show()
print(f"P(Head) from simulation = {heads/1000:.3f}  (theory = 0.5)")
```

## 1️⃣4️⃣ Joint & Conditional — Venn Diagram Style

```python
# 100 students: Cricket, Football, Both
labels  = ['Only Cricket', 'Both', 'Only Football']
values  = [35, 25, 40]
colors  = ['steelblue', 'mediumpurple', 'tomato']

plt.figure(figsize=(7, 4))
plt.bar(labels, values, color=colors, edgecolor='black')
plt.title("Joint & Marginal Probability (100 Students)")
plt.ylabel("Students")
for i, v in enumerate(values):
    plt.text(i, v + 0.5, f'{v/100:.2f}', ha='center', fontweight='bold')
plt.show()

print(f"P(Cricket)        = {(35+25)/100}")
print(f"P(Football)       = {(40+25)/100}")
print(f"P(Both)           = {25/100}")
print(f"P(Cricket|Both)   = {25/(25+40):.3f}")
```

## 1️⃣5️⃣ Bayes' Theorem
$$P(A|B) = \frac{P(B|A) \times P(A)}{P(B)}$$

```python
p_disease = 0.01
p_pos_sick    = 0.95
p_pos_healthy = 0.05

p_pos = p_pos_sick * p_disease + p_pos_healthy * (1 - p_disease)
p_sick_given_pos = (p_pos_sick * p_disease) / p_pos

print(f"P(Disease | Positive) = {p_sick_given_pos:.4f} = {p_sick_given_pos*100:.2f}%")

labels  = ['Disease & Positive', 'No Disease & Positive']
vals    = [p_pos_sick*p_disease, p_pos_healthy*(1-p_disease)]

plt.figure(figsize=(6, 4))
plt.pie(vals, labels=labels, autopct='%1.1f%%',
        colors=['tomato', 'steelblue'], startangle=90)
plt.title("Bayes: Who Tests Positive?")
plt.show()
```

# 📉 PART 8 — Probability Distributions

| Distribution | Type | Use Case |
|---|---|---|
| Uniform | Continuous | Equal chance |
| Normal | Continuous | Exam scores, Heights |
| Binomial | Discrete | Pass/Fail, Coin flip |
| Poisson | Discrete | Calls per hour |

## 1️⃣6️⃣ Uniform Distribution — All Values Equally Likely

```python
data = np.random.uniform(0, 10, 2000)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(data, bins=20, color='coral', edgecolor='black')
axes[0].set_title("Uniform Distribution — Histogram")

sns.kdeplot(data, ax=axes[1], fill=True, color='coral')
axes[1].set_title("Uniform Distribution — KDE")

plt.tight_layout()
plt.show()
```

## 1️⃣7️⃣ Normal Distribution — Bell Curve 🔔
Mean = Median = Mode. 68-95-99.7 Rule applies.

```python
mu, sigma = 70, 10
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 300)
y = norm.pdf(x, mu, sigma)

plt.figure(figsize=(9, 4))
plt.plot(x, y, 'k-', linewidth=2)

# fill std dev bands
plt.fill_between(x, y, where=(x >= mu-sigma) & (x <= mu+sigma),   alpha=0.4, color='steelblue', label='±1 std (68%)')
plt.fill_between(x, y, where=(x >= mu-2*sigma) & (x <= mu+2*sigma), alpha=0.2, color='green',     label='±2 std (95%)')
plt.fill_between(x, y, where=(x >= mu-3*sigma) & (x <= mu+3*sigma), alpha=0.1, color='orange',    label='±3 std (99.7%)')

plt.axvline(mu, color='red', linestyle='--', label=f'Mean = {mu}')
plt.title("Normal Distribution — 68-95-99.7 Rule")
plt.xlabel("Score")
plt.ylabel("Probability Density")
plt.legend()
plt.show()
```

### Verify 68-95-99.7 Rule from Simulation

```python
data = np.random.normal(70, 10, 10000)
m, s = np.mean(data), np.std(data)

w1 = np.mean((data > m-s)   & (data < m+s))   * 100
w2 = np.mean((data > m-2*s) & (data < m+2*s)) * 100
w3 = np.mean((data > m-3*s) & (data < m+3*s)) * 100

print(f"±1 std: {w1:.1f}%   (theory 68%)")
print(f"±2 std: {w2:.1f}%   (theory 95%)")
print(f"±3 std: {w3:.1f}%   (theory 99.7%)")
```

## 1️⃣8️⃣ Binomial Distribution — Success or Failure

```python
n, p = 10, 0.5
x = np.arange(0, 11)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Fair coin (p=0.5)
axes[0].bar(x, binom.pmf(x, 10, 0.5), color='steelblue', edgecolor='black')
axes[0].set_title("Binomial — Fair Coin p=0.5")
axes[0].set_xlabel("Heads in 10 tosses")

# Biased coin (p=0.8)
axes[1].bar(x, binom.pmf(x, 10, 0.8), color='tomato', edgecolor='black')
axes[1].set_title("Binomial — Biased Coin p=0.8")
axes[1].set_xlabel("Heads in 10 tosses")

plt.tight_layout()
plt.show()
```

## 1️⃣9️⃣ Poisson Distribution — Events in Fixed Time

```python
x = np.arange(0, 20)

plt.figure(figsize=(9, 4))
for lam, color in zip([2, 5, 10], ['steelblue', 'tomato', 'green']):
    plt.plot(x, poisson.pmf(x, lam), 'o-', color=color, label=f'λ = {lam}')

plt.title("Poisson Distribution for Different λ")
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.legend()
plt.show()
```

## 2️⃣0️⃣ All Distributions — Side by Side

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

# Uniform
axes[0].hist(np.random.uniform(0,10,1000), bins=20, color='coral',        edgecolor='black')
axes[0].set_title("Uniform")

# Normal
axes[1].hist(np.random.normal(5, 1, 1000),  bins=20, color='steelblue',   edgecolor='black')
axes[1].set_title("Normal")

# Binomial
x = np.arange(0, 11)
axes[2].bar(x, binom.pmf(x, 10, 0.5),       color='mediumseagreen', edgecolor='black')
axes[2].set_title("Binomial")

# Poisson
x2 = np.arange(0, 15)
axes[3].bar(x2, poisson.pmf(x2, 5),         color='mediumpurple',   edgecolor='black')
axes[3].set_title("Poisson")

plt.tight_layout()
plt.show()
```

# ✅ Full Summary

### 📊 Descriptive Statistics
| Topic | Key Concept |
|---|---|
| Mean / Median / Mode | Center of data |
| Variance / Std Dev | Spread of data |
| Z-Score | Position from mean |
| Percentile / Quartile | Rank in data |
| Skewness / Kurtosis | Shape of distribution |
| Correlation | Relationship strength |

### 🎲 Probability
| Topic | Formula |
|---|---|
| Basic | P = favorable ÷ total |
| Joint | P(A∩B) = P(A) × P(B) |
| Conditional | P(A\|B) = P(A∩B) ÷ P(B) |
| Bayes | P(A\|B) = P(B\|A)·P(A) ÷ P(B) |

### 📉 Distributions
| Type | Shape | Example |
|---|---|---|
| Uniform | Flat | Dice |
| Normal | Bell | Exam scores |
| Binomial | Skewed bar | Coin flip |
| Poisson | Right tail | Call centre |
