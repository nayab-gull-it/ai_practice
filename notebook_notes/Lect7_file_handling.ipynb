# File handling

```python
f = open("files/file.txt", 'w')
```

### open , read , write

```python
f.write("Hello from here  ")
f.close()
```

```python
f = open("files/file.txt", 'r' )
print(f.read())
f.close()
```

### with keyword

```python
with open ('files/password.txt' , 'w') as f:
    f.write("pass123")
    
```

```python
with open ('files/password.txt' , 'r') as f:
    print(f.read())
```

### create file 

```python
with open("ai.txt", 'w') as file:
    file.write("AI stands for Artificial Intelligence. It refers to computer systems designed to perform tasks that typically require human intellect—such as learning, reasoning, problem-solving, understanding language, and recognizing patterns. ")
```

```python
with open ("ml.txt", 'w') as file:
    file.write("Machine learning is a subset of artificial intelligence where computers learn from data and identify patterns without being explicitly programmed. ")
with open ("ml.txt", 'r') as file:
    print(file.read())
    file.seek(10)
    file.tell()
```
