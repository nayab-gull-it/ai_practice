## bank account


```python
class BankAccount:

    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    
    def deposit(self, amount):
        self.balance += amount
        print(f"Rs.{amount} deposited successfully.")

   
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Rs.{amount} withdrawn successfully.")
        else:
            print("Insufficient balance!")

    def check_balance(self):
        print(f"Current Balance: Rs.{self.balance}")
```

### Multiple inheritance

```python
class Swimmer:
    """Mixin for swimming behavior."""
    def swim(self):
        return "Can swim in water."

class Flyer:
    """Mixin for flying behavior."""
    def fly(self):
        return "Can fly in the air."

class Duck(Swimmer, Flyer):
    """
    Multiple Inheritance: The 'Duck' class inherits all methods from 
    both 'Swimmer' and 'Flyer' without duplicating logic.
    """
    pass

# Execute
donald = Duck()
print(f"Duck abilities: {donald.swim()} and {donald.fly()}")
```

### Multi-level inheritance

```python
class Vehicle:
    """Grandparent class defining core transport mechanics."""
    def start_engine(self):
        return "Engine started."

class Car(Vehicle):
    """Parent class adding specialized wheeled transport mechanics."""
    def drive(self):
        return f"{self.start_engine()} Driving on roads."

class ElectricCar(Car):
    """
    Multilevel Inheritance: Inherits from 'Car' (Parent) and 'Vehicle' (Grandparent).
    It gains both 'start_engine' and 'drive' capabilities automatically.
    """
    def charge_battery(self):
        return f"{self.drive()} Battery charging active."

# Execute
tesla = ElectricCar()
print(f"EV Status: {tesla.charge_battery()}")
```

### Hierarchical Inheritance

```python
class PaymentProcessor:
    """Parent class defining the base payment protocol."""
    def process(self, amount: float):
        return f"Processing ${amount:.2f}"

class CreditCardPayment(PaymentProcessor):
    """Hierarchical Subclass 1: Specializes the base processor for credit cards."""
    def authorized_charge(self, amount: float):
        return f"{self.process(amount)} securely via Credit Card."

class CryptoPayment(PaymentProcessor):
    """Hierarchical Subclass 2: Specializes the base processor for cryptocurrency."""
    def broadcast_transaction(self, amount: float):
        return f"{self.process(amount)} via Blockchain network."

# Execute
cc, crypto = CreditCardPayment(), CryptoPayment()
print(cc.authorized_charge(150.00))
print(crypto.broadcast_transaction(50.00))
```

### Hybrid Inheritance

```python
class Device:
    """Base Parent Class: Core device properties."""
    def power_on(self):
        return "System powered up."

class Scanner(Device):
    """Hierarchical Branch 1: Specialized scanning asset."""
    def scan(self):
        return "Document digitized."

class Printer(Device):
    """Hierarchical Branch 2: Specialized printing asset."""
    def print_page(self):
        return "Page printed."

class AllInOnePrinter(Scanner, Printer):
    """
    Hybrid Inheritance: Combines Multiple (Scanner, Printer) and Hierarchical (Device).
    Super() tracks MRO seamlessly: AllInOnePrinter -> Scanner -> Printer -> Device.
    """
    def copy_document(self):
        return f"{self.power_on()} {self.scan()} -> {self.print_page()}"

# Execute
office_machine = AllInOnePrinter()
print(office_machine.copy_document())
```
