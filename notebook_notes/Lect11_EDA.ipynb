shape: The dimensions of the array (e. g , number of rows and columns).
size: The total count of items packed inside the array
dtype: The kind of data stored (like whole numbers, decimals, or text).
itemsize: The memory space a single item occupies in bytes.
nbytes: The total memory space the entire array takes up
