```python
import csv

# Student data transcribed from the image
students_data = [
    {"Student ID": "ST001", "Name": "Ali Khan", "Marks": [68, 81, 90, 85]},
    {"Student ID": "ST002", "Name": "Ayesha Noor", "Marks": [95, 89, 92, 94]},
    {"Student ID": "ST003", "Name": "Hamza Ahmed", "Marks": [76, 80, 72, 78]},
    {"Student ID": "ST004", "Name": "Fatima Zahra", "Marks": [91, 93, 95, 90]},
    {"Student ID": "ST005", "Name": "Bilal Hussain", "Marks": [95, 89, 92, 94]}
]

# File path to save the CSV file
csv_file_name = "students_data.csv"

# Writing the data to a .csv file
with open(csv_file_name, "w", newline="", encoding="utf-8") as csv_file:
    # Defining column names
    fieldnames = ["Student ID", "Name", "Marks"]
    
    writer = csv.DictWriter(csv_file, fieldnames=fieldnames)
    
    # Writing headers (column titles)
    writer.writeheader()
    
    # Writing student rows
    for student in students_data:
        # Marks list converted to a clean string format inside the CSV cell
        writer.writerow({
            "Student ID": student["Student ID"],
            "Name": student["Name"],
            "Marks": f"[{', '.join(map(str, student['Marks']))}]"
        })

print(f"CSV data successfully saved to {csv_file_name}")
```
