```python
def mean(val):
    return sum(val)/ len(val)
def median(val):
    val = sorted(val)
    n = len(val)
    mid = n //2

    if n %2 == 0:
        return ( val[mid -1] + val[mid]) / 2
    else: 
        return val[mid]
```

```python
def mode(arr):
    freq = {}

    for num in arr:
        freq[num] = freq.get(num, 0) + 1

    max_count = max(freq.values())

    for key, value in freq.items():
        if value == max_count:
            return key
```
