# function recursion

```python
def fact (x):
    if x==1:
        return 1
    return x * fact(x-1)
```

```python
fact (3)
```

### fibonacci

```python
def fib (n):
    if(n==0 or n==1):
        return 1
    return fib(n-1)+ fib (n-2)

fib(5)
    
        
   
     
```

## Unit converter App

```python
def metre(m):
    feet = 3.2804 * m
    inch = 39.27 * m
    print(f"{m} meters = {feet} ft and {inch} inch")

def feet(f):
    metre = 0.308 * f
    inch = 12 * f
    print(f"{f} feet = {metre} meter and {inch} inch")

def inch(i):
    meter = 0.0254 * i
    feet = 0.083 * i
    print(f"{i} inches = {meter} meter and {feet} feet")
```

```python
while True:
    unit = input("Welcome to Unit Converter\nChoose:\n m for meter\n f for feet\n i for inch\n q for quit\n")

    if unit == 'q' or unit == 'Q':
        break

    val = float(input("Enter a value to convert: "))

    if unit == 'm' or unit == 'M':
        metre(val)
    elif unit == 'f' or unit == 'F':
        feet(val)
    elif unit == 'i' or unit == 'I':
        inch(val)
    else:
        print("Please enter a valid character (m, f, i, q).")
```

```python
import tkinter as tk
from tkinter import ttk

# --- Conversion Logic (Updated to return values instead of printing) ---
def convert_units():
    try:
        val = float(entry_value.get())
        unit = combo_unit.get()
        
        if unit == "Meter (m)":
            feet_val = 3.2804 * val
            inch_val = 39.27 * val
            result_text.set(f"{val} meters =\n{feet_val:.4f} ft\n{inch_val:.4f} inches")
            
        elif unit == "Feet (f)":
            metre_val = 0.308 * val
            inch_val = 12 * val
            result_text.set(f"{val} feet =\n{metre_val:.4f} meters\n{inch_val:.4f} inches")
            
        elif unit == "Inch (i)":
            meter_val = 0.0254 * val
            feet_val = 0.083 * val
            result_text.set(f"{val} inches =\n{meter_val:.4f} meters\n{feet_val:.4f} feet")
            
    except ValueError:
        result_text.set("⚠️ Error: Please enter a valid number!")

# --- UI Setup ---
root = tk.Tk()
root.title("Modern Unit Converter")
root.geometry("400x450")
root.configure(bg="#f3f4f6")  # Light gray background

# Modern Font Styles
FONT_TITLE = ("Helvetica", 16, "bold")
FONT_LABEL = ("Helvetica", 10, "normal")
FONT_RESULT = ("Helvetica", 12, "bold")

# Title Banner
title_label = tk.Label(
    root, text="📐 Unit Converter", font=FONT_TITLE, 
    bg="#3b82f6", fg="white", pady=15
)
title_label.pack(fill="x",  pady=(0, 20))

# Main Container Frame
main_frame = tk.Frame(root, bg="#f3f4f6")
main_frame.pack(padx=20, fill="both", expand=True)

# 1. Input Field
lbl_value = tk.Label(main_frame, text="Enter Value:", font=FONT_LABEL, bg="#f3f4f6", fg="#374151")
lbl_value.pack(anchor="w", pady=(0, 5))

entry_value = ttk.Entry(main_frame, font=("Helvetica", 12))
entry_value.pack(fill="x", ipady=5, pady=(0, 15))
entry_value.focus()

# 2. Dropdown (Combobox) Selector
lbl_unit = tk.Label(main_frame, text="Select Source Unit:", font=FONT_LABEL, bg="#f3f4f6", fg="#374151")
lbl_unit.pack(anchor="w", pady=(0, 5))

combo_unit = ttk.Combobox(
    main_frame, values=["Meter (m)", "Feet (f)", "Inch (i)"], 
    state="readonly", font=("Helvetica", 11)
)
combo_unit.current(0)  # Default to Meter
combo_unit.pack(fill="x", ipady=3, pady=(0, 20))

# Styling the Convert Button using ttk themes
style = ttk.Style()
style.theme_use("clam")
style.configure(
    "TButton", font=("Helvetica", 11, "bold"), 
    background="#10b981", foreground="white", borderwidth=0
)
style.map("TButton", background=[("active", "#059669")])

# 3. Convert Button
btn_convert = ttk.Button(main_frame, text="Convert", command=convert_units, style="TButton")
btn_convert.pack(fill="x", ipady=6, pady=(0, 20))

# 4. Results Card Box
lbl_result_title = tk.Label(main_frame, text="Conversion Results:", font=FONT_LABEL, bg="#f3f4f6", fg="#374151")
lbl_result_title.pack(anchor="w", pady=(0, 5))

result_frame = tk.Frame(main_frame, bg="white", highlightbackground="#e5e7eb", highlightthickness=1, bd=0)
result_frame.pack(fill="both", expand=True, ipady=10)

result_text = tk.StringVar()
result_text.set("Results will appear here...")

lbl_result = tk.Label(
    result_frame, textvariable=result_text, font=FONT_RESULT, 
    bg="white", fg="#1f2937", justify="center"
)
lbl_result.pack(expand=True, fill="both", padx=10, pady=10)

# Bind Enter key to trigger conversion automatically
root.bind('<Return>', lambda event: convert_units())

root.mainloop()
```
