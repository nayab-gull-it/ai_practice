## Lemmatization

```python
nltk.download('wordnet')

from nltk.stem import WordNetLemmatizer

# 2. Lemmatizer initialize karein
lemmatizer = WordNetLemmatizer()

# 3. Jo filtered_words humne stop_words se bachaye thay, unhe lemmatize karein
lemmatized_words = [lemmatizer.lemmatize(w) for w in filtered_words]

print("Lemmatized Words:", lemmatized_words)
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# 1. Thoda bada aur rich text data
text = """
Artificial Intelligence and Machine learning are transforming our world. Data science allows companies 
to extract meaningful insights from massive datasets. Deep learning models, neural networks, and natural 
language processing are driving innovations across healthcare, finance, and education. Developers use 
Python libraries like NumPy, Pandas, Scikit-Learn, and TensorFlow to build smart algorithms. Predictive 
analytics, automated software, data engineering, and cloud computing are essential skills for modern 
tech professionals. Machine learning engineering requires strong knowledge of algorithms, models, and 
data analysis. Future technology will rely heavily on autonomous systems, robotic automation, and 
generative models that can create human-like text and art. Continuous learning and adaptation are key 
in this fast-paced field of artificial intelligence, where new applications and frameworks emerge every day.
"""

# 2. Tokenize aur Lowercase
words = word_tokenize(text.lower())

# 3. Stopwords aur punctuation nikalna
stop_words = set(stopwords.words('english'))
filtered_words = [w for w in words if w not in stop_words and w.isalnum()]

# 4. Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(w) for w in filtered_words]

# 5. Words ko wapas ek bade string mein jodhna
final_text_string = " ".join(lemmatized_words)
print("Data ready hai! Total words jo cloud mein jayenge:", len(lemmatized_words))
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Khubsurat aur bada Word Cloud generate karein
# 'viridis', 'plasma', 'inferno', 'magma', ya 'coolwarm' mein se koi bhi color scheme choose kar sakti hain
wordcloud = WordCloud(
    width=1000, 
    height=600, 
    background_color='white', 
    colormap='viridis',        # Yeh colors ko bohot attractive banata hai
    max_words=100,             # Maximum kitne bade lafz dikhane hain
    min_font_size=10
).generate(final_text_string)

# Screen par display karne ka code
plt.figure(figsize=(12, 7))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis("off")  # Borders hatane ke liye
plt.savefig('my_wordcloud.png', dpi=300, bbox_inches='tight')
plt.show()
```
